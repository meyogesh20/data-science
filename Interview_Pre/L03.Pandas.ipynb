# 1️⃣ Pandas Series

## 📌 Definition

A **Series** is:

> 📍 A **1-dimensional labeled array** capable of storing any data type.

---

## 🧠 Key Characteristics

✔ One-dimensional
✔ Has labels called **Index**
✔ Can store:

* 🔢 Integers
* 🔤 Strings
* 📉 Floats
* ❌ None/NaN
* 🧩 Python objects

---

## 📊 Structure

```python
import pandas as pd

s = pd.Series([10, 20, 30])
```

### Output

| Index | Value |
| ----- | ----- |
| 0     | 10    |
| 1     | 20    |
| 2     | 30    |

---

# 🏷 Important Terms

## 🔹 Values

Actual stored data

```python
[10, 20, 30]
```

## 🔹 Index

Labels associated with data

```python
[0, 1, 2]
```

---

# ✨ Custom Index Example

```python
s = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
```

| Index | Value |
| ----- | ----- |
| a     | 100   |
| b     | 200   |
| c     | 300   |

Access:

```python
s['a']
```

➡ Output: `100`

# 🐼 Pandas Index Objects — Interview Cheat Sheet

# 📌 What is a Pandas Index?

In Pandas, an **Index Object** is used to:

* 🏷 Label rows in a DataFrame
* 🔍 Access data quickly
* ⚡ Enable fast searching & slicing
* 🔄 Align data correctly between DataFrames/Series

Think of Index as:

> 📍 The “address system” of Pandas data.

---

# 🧠 Simple Definition

> An Index is a labeled array used to identify rows/elements in a Series or DataFrame.

Labels can be:

* 🔢 Numbers
* 🔤 Strings
* 📅 Dates/Times

---

# 🚀 Why Index is Important?

✅ Fast lookup
✅ Easy slicing
✅ Better data organization
✅ Automatic data alignment
✅ Efficient joins & merges

---

# 🔥 Key Features of Index Object

## 1️⃣ Immutable

⚠ Once created, index size cannot be changed directly.

```python id="m9dn5j"
df.index[0] = 100   # ❌ Error
```

✔ Index objects are immutable.

---

## 2️⃣ Data Alignment

Pandas automatically aligns data using indexes.

```python id="3n7v20"
s1 + s2
```

➡ Matching indexes are combined automatically.

---

## 3️⃣ Fast Slicing

Efficient label-based access.

```python id="6lbv6z"
df.loc[0:5]
```

---

# 🏗 The Index Class

## 📌 Syntax

```python id="mwv0b7"
pandas.Index(
    data=None,
    dtype=None,
    copy=False,
    name=None,
    tupleize_cols=True
)
```

---

# 🧩 Parameters Explained

| Parameter       | Meaning                       |
| --------------- | ----------------------------- |
| `data`          | Data for index                |
| `dtype`         | Data type                     |
| `copy`          | Create copy or not            |
| `name`          | Name of index                 |
| `tupleize_cols` | Create MultiIndex if possible |

---

# 📊 Example of Index

```python id="3cuhmn"
import pandas as pd

idx = pd.Index([101, 102, 103], name="EmpID")

print(idx)
```

### Output

```python id="mk4x4v"
Index([101, 102, 103], dtype='int64', name='EmpID')
```
