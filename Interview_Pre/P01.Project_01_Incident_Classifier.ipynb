## **Project: Incident Classifier & Historical Incident Recommender**

Developed an AI-driven NLP solution to automate ticket classification and recommendation within a global enterprise helpdesk system. The existing manual process relied on a 350+ rule–based cheat sheet, making ticket assignment slow, error-prone, and heavily dependent on support staff experience. To eliminate manual lookup and incorrect routing, built an NLP-based classification model that predicts the correct ticket queue based on the user’s issue description and recommends similar historical incidents with closing notes to accelerate resolution.

### **Key Features**

* Implemented an NLP-driven ticket classifier that predicts the appropriate support queue directly from the ticket description entered by the user.
* Built a historical incident recommender using text embeddings to surface similar past tickets along with resolution notes for faster troubleshooting.
* Automated learning pipeline ensures that new ticket patterns and updates are continuously learned by the model, removing dependency on manual cheat-sheet maintenance.

### **Business Impact**

* Reduced ticket assignment effort for the service desk from **12% to 5%**.
* Reduced first-time incorrect assignments from **~20% to ~5%**.
* Eliminated the need for maintaining 350+ rule cheat sheets and frequent training sessions for new support analysts.
* Improved first-call resolution by enabling analysts to quickly reference similar historical incidents and their closing notes.

# ✅ **FIVE-MINUTE DEEP TECHNICAL EXPLANATION**

### **1. Problem Background**

Our enterprise helpdesk received thousands of global support tickets daily. Ticket assignment was completely manual and depended on a 350+ rule cheat sheet.
This created major issues:

* ~20% first-time wrong assignment
* Ticket hopping between teams
* High onboarding time for new agents
* Rules becoming stale and difficult to maintain

We needed a scalable, learning-based system to improve accuracy and provide faster resolution support.

---

### **2. Data Understanding & Preparation**

I extracted several years of historical ticket data from **SQL and MongoDB**—including ticket descriptions, assigned queues, and closing notes.

**Data cleaning steps:**

* Text normalization, tokenization, lemmatization using **spaCy**
* Removal of noise, signatures, and system-generated text
* Class distribution checks and balancing (oversampling & class weighting)
* Train/validation/test split ensuring no temporal leakage

This created a stable dataset for training the classification and similarity models.

---

### **3. NLP Classification Model**

We evaluated classical ML vs modern NLP:

| Approach                           | Notes                                                   |
| ---------------------------------- | ------------------------------------------------------- |
| SVM, XGBoost                       | Good baseline but struggled with semantic understanding |
| **Transformers (BERT/DistilBERT)** | Best semantic performance                               |

**Final approach:**
A **fine-tuned Transformer model** using HuggingFace, trained on multi-class queue classification.

**Why?**

* Captures context
* Robust to noisy text
* Adapts to new ticket types

**Training pipeline:**

* Implemented using **PyTorch** + Transformers
* Optimizer: AdamW
* Learning rate scheduling
* Used **cross-entropy loss** with class weights

**Performance:**
High precision/recall across dominant queues, achieving business-grade reliability.

---

### **4. Historical Incident Recommender**

For faster ticket resolution, I created a similarity search engine.

**Steps:**

1. Generated embeddings using **Sentence Transformers**.
2. Stored vectors in **FAISS / Pinecone / ChromaDB**.
3. Applied **Approximate Nearest Neighbor (ANN)** search.
4. Retrieved top-k similar tickets + closing notes.

This allowed support agents to instantly see:

* How similar incidents were resolved
* Steps taken by other teams
* Patterns for recurring issues

This drastically cut down troubleshooting time.

## **1. Why did you choose Transformers instead of traditional ML models?**

**Answer:**
Traditional models like SVM, Logistic Regression, or XGBoost rely heavily on TF-IDF or bag-of-words features, which fail to capture semantic meaning in ticket descriptions.
Transformers, especially BERT-based models, understand context, synonyms, domain-specific phrasing, and long descriptions. In our case, they improved queue classification accuracy by a significant margin and generalized better to new unseen ticket types.

---

## **2. How did you handle imbalance across different ticket queues?**

**Answer:**
I used a combination of techniques:

* Applied **class weights** during loss computation
* Minor oversampling of rare queues
* Stratified train-test split
  This ensured that rare but critical categories were learned properly without overfitting.

---

## **3. How does the recommender system actually retrieve similar tickets?**

**Answer:**

1. Convert ticket descriptions into embeddings using **Sentence Transformers**.
2. Store embeddings in a **vector database** such as FAISS or Pinecone.
3. Perform **ANN (Approximate Nearest Neighbor)** search to retrieve top-k semantic matches.
4. Return matching incidents with closing notes for reference.

This ensures fast retrieval even with millions of tickets.

## **8. What is the difference between OOTB BERT and a fine-tuned model?**

**Answer:**
OOTB BERT provides generic language understanding.
Fine-tuned BERT learns domain-specific patterns from ticket descriptions, such as:

* product names
* error codes
* internal system terminology

This improved queue prediction accuracy significantly.

## **9. Why did you choose FAISS/Pinecone instead of a SQL LIKE search?**

**Answer:**
SQL LIKE patterns perform string matching, not semantic matching.
Vector databases compute similarity based on meaning, not characters.

Results:

* Much higher accuracy
* O(1) ANN search for millions of tickets
* Better scalability and lower latency
