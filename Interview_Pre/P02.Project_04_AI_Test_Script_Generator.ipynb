### **Project Summary**

Built an end-to-end **NLP-powered automated test-script generator** that converts human-written test steps into executable automation code. Designed the architecture, trained domain-specific models, engineered a semantic retrieval pipeline, and integrated it with an existing test framework.

---

### **Key Responsibilities & Achievements**

#### **🔹 1. Requirement Understanding & Dataset Engineering**

* Reverse-engineered ~1000 existing test cases and their execution logs to build a **domain-specific training corpus**.
* Created structured datasets with **action → function mapping**, parameter extraction rules, and automation framework metadata.
* Designed a custom annotation schema for steps, actions, objects, conditions, and expected results.

#### **🔹 2. NLP Pipeline & Semantic Understanding**

* Built a multi-stage natural language understanding pipeline including:

  * Sentence segmentation & cleaning
  * POS-tagging & dependency parsing (spaCy)
  * Action-object extraction
  * Parameter identification using regex + statistical tagging
* Implemented a **semantic similarity layer** using Sentence Transformers to match user steps to closest automation functions.

#### **🔹 3. Retrieval-Augmented Code Generation**

* Implemented a LangChain-based retrieval module using embeddings for:

  * Function discovery
  * Step interpretation
  * Template selection
* Added rule-based validation to ensure correct argument mapping, sequence alignment, and pre-conditions before generating code.

#### **🔹 4. Code Generator & Framework Integration**

* Designed a robust code generator that:

  * Converts interpreted steps into **ready-to-run test scripts**
  * Inserts framework imports, boilerplate, and function calls
  * Handles dynamic waits, error handling, and assertions
* Integrated the generator with an existing automation framework (UFT/Java), enabling consistent code structure.

#### **🔹 5. Quality, Evaluation & Iterative Improvement**

* Built an automated evaluation loop to compare generated scripts vs. ground truth scripts using:

  * Token-level similarity
  * Function call topology
  * Execution validation
* Achieved **~82% automation accuracy** in initial versions and improved to **~94%** after iterative fine-tuning.

#### **🔹 6. Tooling, Documentation & Collaboration**

* Created CLI tool + UI prototype to allow testers to input natural language test steps.
* Wrote complete documentation covering design decisions, architecture, and usage guidelines.
* Collaborated with QA teams to refine test style guidelines to improve LLM understanding.

---

### **Impact**

* Reduced test script development time by **70–80%**.
* Enabled non-technical QA members to create automation-ready scripts.
* Established reusable NLP + retrieval architecture for future automation tasks.
* Significantly improved consistency, maintainability, and coverage of automated tests.

# **1. Explain your architecture in detail. How does the system work end-to-end?**

### **Answer:**

The system follows a modular NLP → Retrieval → Code Generation pipeline:

1. **NLP Parsing Layer**

   * Uses spaCy dependency parsing, POS tagging, and custom rules to extract:

     * Action (verb)
     * Object (UI element)
     * Parameters (data values)
   * Handles multi-intent sentences by splitting on verbal roots.

2. **Embedding Layer**

   * Converts each test step into a dense vector using **Sentence-BERT**.
   * Automation framework functions are also embedded.

3. **Vector Retrieval Layer**

   * Uses **FAISS** (Flat or HNSW index) for fast Top-K semantic retrieval.
   * Candidate functions based on cosine similarity.

4. **Ranking & Mapping Layer**

   * Function candidates are re-ranked by:

     * Action-object match score
     * Parameter availability
     * Contextual penalties
   * Validates function signatures and maps extracted parameters.

5. **Code Generator**

   * Converts selected function + parameters into final code template.
   * Performs AST-based validation to ensure syntactic correctness.

6. **Evaluation & Execution**

   * Generated script is validated and optionally test-run for correctness.

# **2. Why did you choose Sentence-BERT for embeddings? How did you evaluate embedding quality?**

### **Answer:**

I needed an embedding model that captured semantic meaning of short task-oriented instructions.

**Why Sentence-BERT:**

* Optimized for semantic similarity
* Outperforms BERT in sentence-level retrieval tasks
* Much faster due to cosine similarity–friendly embedding space

**Quality Evaluation:**

* Computed **cosine similarity distributions** between:

  * Positive pairs (correct step–function)
  * Negative pairs
* Measured **ROC-AUC** of similarity threshold
* Performed **manual qualitative checks** on retrieved functions
* Post-tuning accuracy improved from **68% → 92%**

# **3. Explain how FAISS indexing works and why you selected the index type you used.**

### **Answer:**

FAISS supports multiple index types:

* **Flat Index** – brute-force, highest recall, slower for very large datasets
* **HNSW** – graph-based, logarithmic search time
* **IVF Flat** – inverted file, speeds up search for large corpora

I used:

* **Flat Index** during development for perfect recall
* **HNSW** in production for:

  * low-latency lookup
  * high recall balance
  * dynamic index updates

FAISS was chosen because it gives:

* GPU acceleration
* Stable ANN search
* Industry-grade scalability

# **4. How do you handle multi-intent or ambiguous test steps?**

### **Answer:**

Ambiguity like “Enter username and click login” has two intents.

I handle it using a **three-step method**:

1. **Dependency Parsing**

   * Extract all verbs as potential actions
   * Create subtrees for each action phrase

2. **Intent Classification**

   * Classify each sub-step separately
   * Maintain context from previous steps

3. **Semantic Retrieval**

   * Search for functions per action
   * Parameter mapping done independently per sub-step

This ensures multi-intent steps are decomposed and mapped correctly.

# **5. How does your ranking layer score and select the correct function?**

### **Answer:**

Ranking is based on a weighted scoring system:

**Final Score = α(similarity) + β(action_match) + γ(object_match) – δ(parameter_mismatch)**

* **Similarity score:** cosine similarity from Sentence-BERT
* **Action match:** verb normalization → penalty for mismatched actions
* **Object match:** noun similarity & UI element dictionary
* **Parameter check:** incomplete parameter mapping reduces confidence

The candidate with highest composite score > threshold gets selected.

# **6. How exactly do you extract parameters from the user’s natural language step?**

### **Answer:**

Parameter extraction uses:

1. **Dependency Parsing**

   * Identify direct objects, numeric literals, quoted strings

2. **POS Tagging**

   * Noun chunks for object names
   * ADJ + NOUN patterns for attribute values

3. **Regex Rules**

   * Inputs like emails, dates, numbers, strings
   * UI element patterns like “textbox”, “dropdown”

4. **Fallback: semantic expansion**

   * If missing, we infer from similar historical steps

This hybrid approach gives **high recall and high precision**.

# **12. Describe your fine-tuning process. What loss function did you use?**

### **Answer:**

Used **Siamese fine-tuning** for Sentence-BERT:

* **Loss:** Triplet loss (anchor, positive, hard-negative)
* **Why:** Encourages the model to cluster correct pairs close together while pushing wrong pairs apart
* **Training strategy:**

  * batch-hard negative mining
  * cosine similarity margin = 0.3
  * learning rate warm-up + linear decay
