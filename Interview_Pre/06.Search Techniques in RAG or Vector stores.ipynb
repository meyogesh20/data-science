# Different Search Techniques Used in Vector Retrieval / RAG Systems

In LangChain, FAISS, pgvector, and modern RAG systems, retrieval quality depends heavily on the search strategy used.

These search techniques are mainly designed to optimize:

* Relevance
* Speed
* Context quality
* Scalability
* Recall and precision

---

# 1. Dense Vector Similarity Search

### What it is

Converts text into embeddings and retrieves nearest vectors.

### Works Using

* Cosine similarity
* Dot product
* Euclidean distance

### Flow

Query → Embedding → Compare with stored vectors → Top-K retrieval

### Best For

* Semantic understanding
* Natural language retrieval

### Example

Query:

> “How to reset password?”

Can retrieve:

> “Steps for changing login credentials”

even without exact keywords.

### Common Tools

* FAISS
* Milvus
* Pinecone
* pgvector

---

# 2. Sparse Retrieval (Keyword Search)

### What it is

Traditional token/keyword-based retrieval.

### Algorithms

* BM25
* TF-IDF

### Best For

* Exact keywords
* IDs
* Codes
* Error messages
* Legal terms

### Example

Searching:

> “ORA-00904”

needs exact keyword retrieval.

### Pros

* High precision for exact terms

### Cons

* No semantic understanding

### Common Tools

* Elasticsearch
* OpenSearch

---

# 3. Hybrid Search

### What it is

Combines:

* Dense retrieval
* Sparse retrieval

### Formula

Final Score =
Semantic Score + Keyword Score

### Why Important

Dense search may miss exact keywords.
Sparse search may miss semantic meaning.

Hybrid combines both.

### Best For

Enterprise RAG systems.

### Example

Query:

> “Java NullPointerException in payment service”

Needs:

* semantic understanding
* exact exception matching

### Enterprise Usage

Most production RAG systems use hybrid retrieval.

### Common Systems

* OpenSearch
* Elasticsearch
* Weaviate

---

# 4. Approximate Nearest Neighbor (ANN) Search

### What it is

Fast approximate vector retrieval.

Instead of exact comparison with all vectors:

* intelligently narrows search space

### Why Needed

Exact search becomes slow for millions/billions of vectors.

### Tradeoff

* Slightly lower accuracy
* Massive speed improvement

### Best For

Large-scale enterprise systems.

# Interview One-Line Summary

> Modern enterprise RAG systems typically use hybrid retrieval combining dense semantic search, sparse keyword search, ANN indexing like HNSW, metadata filtering, and reranking pipelines to achieve high recall, precision, scalability, and low latency.


In a production-level RAG or vector search architecture, relying solely on basic k-Nearest Neighbors (k-NN) vector lookups is rarely enough. Real-world user queries are complex, data can be messy, and latency constraints require highly optimized search strategies.

Production RAG systems categorize search into five core areas: **Exact vs. Approximate Search**, **Hybrid Approaches**, **Query Pre-processing Search**, **Advanced Retrieval Structures**, and **Post-Retrieval reranking**.

---

## 1. Exact vs. Approximate Vector Search (The Core Indexing Layers)

Before searching, the vector database sets up an index. The choice of index dictates how the search is executed mathematically.

* **Flat Index (Exact Search / k-NN):**
* **How it works:** It calculates the distance (Cosine, Euclidean, or Dot Product) between the query vector and *every single vector* in the database.
* **When to use:** Small datasets ($<10,000$ vectors) where $100\%$ accuracy (recall) is required. It scales linearly $O(N)$, making it too slow for massive production datasets.


* **HNSW (Hierarchical Navigable Small World - Approximate Search / ANN):**
* **How it works:** It creates a multi-layered graph structure, similar to a skip-list. The top layers have sparse connections for fast, long-range routing across the data space, while the bottom layers contain dense connections for granular local pinpointing.
* **When to use:** The production standard for ultra-low latency ($O(\log N)$ search time) with negligible loss in accuracy.


* **IVF (Inverted File Index):**
* **How it works:** It uses k-means clustering to partition the entire vector space into Voronoi cells. During a search, the system identifies the closest cluster centroids to the query and only searches within those specific cells.
* **When to use:** When RAM is constrained. It drastically reduces the search footprint but can miss edge-case vectors if too few clusters are checked.



---

## 2. Hybrid Search (Dense + Sparse)

This is a mandatory pattern for enterprise search engines to avoid semantic "hallucinations" on highly specific terms.

* **Dense Retrieval (Vector Search):** Uses deep learning models (like OpenAI or BGE) to capture the **semantic meaning and context** of a query. It understands that "automobile" and "car" are the same thing.
* **Sparse Retrieval (Keyword Search / BM25):** Uses statistical frequency counting (like TF-IDF or BM25) to look for **exact keyword matches**. It doesn’t understand context, but it excels at searching for serial numbers, product IDs, SKU codes, or highly specific error codes (e.g., `NullPointerException`).
* **The Fusion Layer:** A production pipeline executes both searches in parallel and blends the results using algorithms like **Reciprocal Rank Fusion (RRF)** or normalized score distribution.

---

## 3. Query Pre-Processing Search Techniques

Instead of passing the raw user query directly to the vector store, the system transforms the query first to maximize retrieval success.

* **Query Deconstruction / Sub-Query Generation:** If a user asks a complex question ("Compare our revenue in Q1 vs Q2"), an LLM breaks it down into independent search items ("Q1 revenue figures" and "Q2 revenue figures"), executes separate vector searches, and pools the documents.
* **Query Rewriting / Expansion:** Users often phrase queries poorly. The system uses a fast, lightweight LLM to rewrite the query into multiple variations or inject synonyms before generating the query embedding.
* **Hypothetical Document Embeddings (HyDE):**
* **How it works:** The system takes the user's query and asks an LLM to generate a *fake, hypothetical answer*. It then embeds that fake answer and uses *it* to search the vector database.
* **Why it works:** Vector models are trained to find similarity between documents. A fake answer is structurally closer to the real answer in vector space than a short question is.



---

## 4. Advanced Structural Retrieval Techniques

These methods structure how documents are split and retrieved to maintain maximum context for the LLM generation phase.

* **Parent-Child / Small-to-Large Retrieval:**
* **How it works:** Raw documents are split into tiny chunks (e.g., 100-200 tokens) for highly precise vector matching. However, each tiny chunk is linked to a larger "parent" chunk (e.g., 1000 tokens) or the full document context.
* **Result:** The system searches using the precise small chunk but passes the rich, larger parent chunk to the LLM.


* **Hierarchical Summary Retrieval:** Instead of indexing raw text chunks directly, an LLM generates a clean summary for each document chapter. The summaries are embedded and searched first. Once a summary matches, the system drills down into the raw, granular chunks nested under that summary.
* **Metadata Filtering (Pre-filtering vs. Post-filtering):** Hard filters applied alongside semantic search (e.g., `WHERE department == 'HR' AND year >= 2025`). Production architectures use **Pre-filtering** (filtering the index *before* calculating vector distances) to prevent accuracy degradation.

---

## 5. Post-Retrieval Reranking (The Final Gate)

Vector databases optimize for speed, meaning the top 20 documents returned by an ANN search might not be in the absolute best semantic order.

* **Cross-Encoder Rerankers (e.g., Cohere Rerank, BGE-Reranker):**
* **How it works:** While initial vector search uses Bi-Encoders (embedding queries and documents independently), a Reranker is a **Cross-Encoder**. It takes the query and a retrieved document *together* and processes them simultaneously through a transformer model to compute a highly accurate relevance score.
* **Production Setup:** You retrieve a wide net of documents (e.g., Top 50) very quickly using HNSW vector search, then pass those 50 documents through a Reranker to select the absolute best Top 5 to send to your LLM prompt. This radically slashes the "needle in a haystack" problem.



---

### How to summarize this for an Interview

> *"In production RAG systems, we view retrieval as a multi-stage pipeline rather than a single vector lookup. We utilize **HNSW indexing** for low-latency approximate nearest neighbor search, but we always pair it with a **BM25 sparse search layer** via Reciprocal Rank Fusion (RRF) to catch exact product codes and keywords. *
> *To maximize accuracy, we implement query manipulation techniques like **HyDE** or Sub-Query generation on the front end, use a **Small-to-Large (Parent-Child) token mapping** strategy within our vector database to preserve context, and finally gate our results through a **Cross-Encoder Reranker** to filter out semantic noise before hitting the LLM context window."*

**BM25 (Best Matching 25)**—often referred to as Okapi BM25—is the gold standard algorithm for **sparse, keyword-based search**. It is a statistical formula used by search engines (like Elasticsearch, OpenSearch, and Lucene) to rank a set of documents based on the matching keywords in a user's query.

Even in the era of advanced AI and Vector Search, BM25 remains a mandatory component in production RAG systems because it does something vector embeddings struggle with: **it finds exact matches for specific keywords, serial numbers, product IDs, and rare technical jargon.**

---

## 1. The Core Evolution: From TF-IDF to BM25

To understand BM25, you have to understand its predecessor, **TF-IDF (Term Frequency-Inverse Document Frequency)**. BM25 is essentially a highly optimized, non-linear upgrade to TF-IDF.

Both algorithms rely on two main concepts:

1. **Term Frequency (TF):** How often does a search word appear in a document? If a document mentions "Kubernetes" 10 times, it's likely highly relevant.
2. **Inverse Document Frequency (IDF):** How unique is this search word across the *entire dataset*? Words like "the", "is", or "database" appear everywhere, so they get a low weight. Rare words like "Kafka" get a very high weight.

### The Fatal Flaw of TF-IDF

In standard TF-IDF, Term Frequency scales **linearly**. If a document mentions "Kubernetes" 5 times, it gets a score. If it mentions it 100 times (perhaps due to keyword stuffing), its score skyrockets 20 times higher. This ruins search quality.

### How BM25 Fixes It

BM25 introduces **Term Frequency Saturation** and **Document Length Normalization**.

---

## 2. The BM25 Mathematical Formula (Deconstructed)

Don't let the math intimidate you; it breaks down into three distinct, logical components. For a query $Q$ containing keywords $q_1, q_2, \dots, q_n$, the BM25 score for a specific document $D$ is calculated as:

$$\text{Score}(D, Q) = \sum_{i=1}^{n} \text{IDF}(q_i) \cdot \frac{f(q_i, D) \cdot (k_1 + 1)}{f(q_i, D) + k_1 \cdot \left(1 - b + b \cdot \frac{|D|}{\text{avgdl}}\right)}$$

Let's break down the three parts that control this equation:

### Component A: The IDF Component (The Rarity Scale)

$$\text{IDF}(q_i) = \ln \left( \frac{N - n(q_i) + 0.5}{n(q_i) + 0.5} + 1 \right)$$

* $N$: The total number of documents in your database.
* $n(q_i)$: The number of documents containing the specific keyword $q_i$.
* **What it does:** If a word appears in almost every document ($n(q_i) \approx N$), the fraction drops, making the IDF close to 0. If it appears in only 2 documents out of a million, the IDF skyrockets.

### Component B: Term Frequency Saturation (Controlled by $k_1$)

$$\frac{f(q_i, D) \cdot (k_1 + 1)}{f(q_i, D) + k_1 \cdot (\dots)}$$

* $f(q_i, D)$: The frequency of the keyword inside document $D$.
* $k_1$: A tuning parameter (typically set between **1.2 and 2.0**).
* **What it does:** It puts a ceiling on term frequency. As a word appears more and more, the score contribution curves off and flattens out (saturates).
* *Example:* Seeing a keyword 2 times instead of 1 time gives a big score boost. Seeing it 50 times instead of 49 times gives almost zero additional boost.



### Component C: Document Length Normalization (Controlled by $b$)

$$1 - b + b \cdot \frac{|D|}{\text{avgdl}}$$

* $|D|$: The length of the current document (word count).
* $\text{avgdl}$: The average length of all documents in the database.
* $b$: A tuning parameter (typically set to **0.75**).
* **What it does:** It penalizes long, wordy documents. If a 100-page book and a 1-page essay both mention the word "Bitcoin" 3 times, the 1-page essay is obviously much more explicitly *about* Bitcoin. The formula divides by the document length relative to the average length to level the playing field.
* If $b=1$, full length normalization is applied (long documents are heavily penalized).
* If $b=0$, length normalization is completely turned off.



---

## 3. Why BM25 is Crucial for Production RAG (Dense vs. Sparse)

In a production RAG system, relying *only* on Vector Search (Dense Retrieval) introduces massive blind spots. Here is why you must pair it with BM25 in a Hybrid Search architecture:

| Feature | Dense Vector Search (e.g., OpenAI / HNSW) | Sparse Search (BM25) |
| --- | --- | --- |
| **How it searches** | Conceptual meaning & Context | Exact character match |
| **Strengths** | Synonyms, themes, answering vague questions. | Product IDs, Error Logs, SKU Numbers, unique names. |
| **Weaknesses** | Fails completely on exact specific strings or IDs. | Blind to synonyms ("auto" vs "car" yields 0 match). |
| **Compute Cost** | High (Requires GPUs + Vector Database indices) | Very Low (Inverted Index on CPU) |

### The Production Pattern: Hybrid Search + RRF

Because of these complementary strengths, production systems run **BM25** and **Vector Search** side-by-side:

```
                  ┌──> BM25 (Sparse) ───> Top 50 Docs ──┐
User Query ───────┤                                     ├──> Reciprocal Rank Fusion (RRF) ──> Top 5 Docs ──> LLM
                  └──> Vectors (Dense) ─> Top 50 Docs ──┘

```

The results are blended using **Reciprocal Rank Fusion (RRF)**, a scoring system that re-ranks documents based on how high they appeared in *both* the BM25 and Vector search results.

---

## 4. How to Ace This in an Interview

If an interviewer asks you to explain BM25, structure your answer like this to demonstrate deep engineering expertise:

> *"BM25 is the industry-standard sparse retrieval algorithm based on a non-linear optimization of TF-IDF. It scores document relevance using three key mechanisms: **IDF** to calculate word rarity, **Term Frequency Saturation** controlled by the $k_1$ parameter to prevent keyword stuffing from skewing results, and **Document Length Normalization** controlled by the $b$ parameter to ensure short, concise documents aren't penalized against long, wordy files.*
> *In production RAG architectures, we never deploy Vector Search alone. We implement a Hybrid Search strategy where BM25 acts as our safety net. While the Vector database handles semantic, abstract queries, BM25 guarantees that if a user searches for an exact alphanumeric asset ID, a specific error log code like `Err_Code_404`, or a strict system configuration parameter, the system will retrieve the exact document containing that token with near-zero latency."*

# Keyword Search in RAG / Retrieval Systems

Keyword search is a traditional retrieval technique where documents are retrieved based on **exact word or token matching** between the user query and stored documents.

Instead of understanding meaning like vector search, keyword search focuses on:

* exact terms
* token frequency
* relevance scoring
* textual matching

---

# Simple Example

## Query

```text
ORA-00904 invalid identifier
```

## Retrieved Document

```text
Oracle SQL error ORA-00904 occurs when...
```

This works very well because the exact keyword exists.

---

# Core Idea

Keyword search answers:

> “Which documents contain the same or similar words as the query?”

It does NOT understand semantic meaning.

---

# How Keyword Search Works

## Step 1 — Tokenization

Sentence is broken into tokens.

Example:

```text
"Python is powerful"
```

becomes:

```text
["python", "is", "powerful"]
```

---

## Step 2 — Indexing

System creates an inverted index.

---

# Inverted Index

Most important concept in keyword search.

It maps:

```text
word → document IDs
```

Example:

| Word     | Documents |
| -------- | --------- |
| python   | D1, D3    |
| java     | D2        |
| database | D1, D2    |

---

# Query Execution

If user searches:

```text
python database
```

Search engine finds documents containing:

* python
* database

Then ranks them.

# Common Keyword Search Algorithms

---

# 1. TF-IDF

TF-IDF =
Term Frequency × Inverse Document Frequency

---

## A. Term Frequency (TF)

Measures:

> How often word appears in document

Formula:

TF(t,d)=\frac{\text{Count of term }t\text{ in document }d}{\text{Total terms in document}}

Example:
If “Python” appears 5 times,
higher TF score.

---

## B. Inverse Document Frequency (IDF)

Measures:

> How unique a word is across documents

Formula:

IDF(t)=\log\left(\frac{N}{df(t)}\right)

Where:

* N = total documents
* df(t) = documents containing term t

---

## Why Important

Common words like:

* the
* is
* and

get lower importance.

Rare words get higher importance.

---

# TF-IDF Problem

It does not consider:

* word meaning
* context
* semantics

---

# 2. BM25 (Most Popular)

BM25 is an improved version of TF-IDF.

Used in:

* Elasticsearch
* OpenSearch
* Lucene

---

# BM25 Features

It considers:

* term frequency
* document length normalization
* inverse document frequency

---

# BM25 Formula

BM25(D,Q)=\sum_{i=1}^{n} IDF(q_i)\cdot\frac{f(q_i,D)(k_1+1)}{f(q_i,D)+k_1\left(1-b+b\cdot\frac{|D|}{avgdl}\right)}

---

# Why BM25 Better Than TF-IDF

BM25 handles:

* long documents
* keyword saturation
* relevance balancing

much better.

# Interview One-Line Summary

> Keyword search is a lexical retrieval technique that uses token matching, inverted indexes, and ranking algorithms like BM25 or TF-IDF to retrieve documents based on exact word occurrence rather than semantic meaning.


```python
# pip install langchain langchain-community langchain-chroma chromadb sentence-transformers rank_bm25

from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_community.embeddings import HuggingFaceEmbeddings
from rank_bm25 import BM25Okapi
import numpy as np

# ---------------------------------------------------
# Sample Documents
# ---------------------------------------------------

documents = [
    Document(
        page_content="Python is widely used for AI and machine learning."
    ),
    Document(
        page_content="Java is commonly used in enterprise backend systems."
    ),
    Document(
        page_content="ChromaDB is a vector database used in RAG systems."
    ),
    Document(
        page_content="BM25 is a keyword-based ranking algorithm."
    ),
    Document(
        page_content="Cosine similarity is used for semantic vector search."
    ),
]

# ---------------------------------------------------
# Embedding Model
# ---------------------------------------------------

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# ---------------------------------------------------
# Create Chroma Vector DB
# ---------------------------------------------------

vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embedding_model,
    persist_directory="./chroma_db"
)

# ---------------------------------------------------
# BM25 Setup
# ---------------------------------------------------

# Tokenize documents for BM25
tokenized_docs = [
    doc.page_content.lower().split()
    for doc in documents
]

bm25 = BM25Okapi(tokenized_docs)

# ---------------------------------------------------
# User Query
# ---------------------------------------------------

query = "What is vector search in RAG?"

# ---------------------------------------------------
# BM25 Retrieval
# ---------------------------------------------------

tokenized_query = query.lower().split()

bm25_scores = bm25.get_scores(tokenized_query)

print("\n========== BM25 RESULTS ==========\n")

for idx, score in enumerate(bm25_scores):
    print(f"Doc {idx+1}")
    print(f"BM25 Score: {score:.4f}")
    print(documents[idx].page_content)
    print()

# ---------------------------------------------------
# Cosine Similarity Retrieval using ChromaDB
# ---------------------------------------------------

print("\n========== VECTOR SEARCH RESULTS ==========\n")

vector_results = vectorstore.similarity_search_with_score(
    query=query,
    k=3
)

for doc, score in vector_results:
    print(f"Cosine Distance Score: {score:.4f}")
    print(doc.page_content)
    print()

# ---------------------------------------------------
# Hybrid Retrieval (BM25 + Vector Search)
# ---------------------------------------------------

print("\n========== HYBRID SEARCH ==========\n")

# Normalize BM25 scores
bm25_norm = bm25_scores / np.max(bm25_scores)

# Get embeddings manually
query_embedding = embedding_model.embed_query(query)

doc_embeddings = embedding_model.embed_documents(
    [doc.page_content for doc in documents]
)

# Cosine Similarity Calculation
cosine_scores = []

for emb in doc_embeddings:

    emb = np.array(emb)
    query_emb = np.array(query_embedding)

    cosine_similarity = np.dot(query_emb, emb) / (
        np.linalg.norm(query_emb) * np.linalg.norm(emb)
    )

    cosine_scores.append(cosine_similarity)

cosine_scores = np.array(cosine_scores)

# Hybrid Score
# Weighted combination

hybrid_scores = (
    0.5 * bm25_norm +
    0.5 * cosine_scores
)

# Sort documents by hybrid score
ranked_indices = np.argsort(hybrid_scores)[::-1]

for rank, idx in enumerate(ranked_indices):

    print(f"Rank {rank+1}")
    print(f"Hybrid Score: {hybrid_scores[idx]:.4f}")
    print(f"BM25 Score: {bm25_scores[idx]:.4f}")
    print(f"Cosine Similarity: {cosine_scores[idx]:.4f}")
    print(documents[idx].page_content)
    print()
```


---

# Why This is Important in Enterprise RAG

Pure vector search may fail for:

* error codes
* IDs
* exact terms

Pure BM25 may fail for:

* semantic meaning
* conversational queries

Hybrid search gives:

* better recall
* better precision
* enterprise-grade retrieval quality

---

# Production Improvements

In real enterprise systems:

* Use HNSW indexing
* Add reranking
* Use Reciprocal Rank Fusion (RRF)
* Add metadata filtering
* Store embeddings in persistent vector DB
* Use async retrieval
* Add contextual compression

---

# Enterprise Stack Example

| Layer            | Technology       |
| ---------------- | ---------------- |
| Sparse Retrieval | BM25             |
| Dense Retrieval  | ChromaDB / FAISS |
| Embeddings       | BGE / OpenAI     |
| Reranker         | Cohere Rerank    |
| Vector DB        | Chroma           |
| Framework        | LangChain        |


## Child-Parent Retrieval

In LangChain, **Child-Parent Retrieval** (often implemented via the `ParentDocumentRetriever`) 
is a highly effective strategy designed to solve a classic RAG (Retrieval-Augmented Generation) dilemma:

* You want **small chunks** (children) so the embedding model can capture precise semantic meaning and find exact matches.
* You want **large chunks** (parents) passed to the LLM so it has enough surrounding context to generate an accurate, coherent answer.

Here is a breakdown of the strategy, how storage works, and how to implement it.

---

## The Strategy: How It Works

Instead of storing documents and retrieving them as-is, the Parent Document Retriever splits your data into two tiers:

1. **Parent Documents:** The larger, context-rich chunks (or even the entire original documents).
2. **Child Documents:** Smaller sub-chunks split directly from the parents.

### The Retrieval Flow

1. You embed and store only the **child chunks** in a Vector Database.
2. When a user asks a query, the system searches the Vector Database and finds the most relevant **child chunks**.
3. Instead of giving those tiny child chunks to the LLM, the system looks up the **unique Parent ID** attached to those children and fetches the larger **parent chunk** from a separate document store.
4. The LLM receives the full context of the parent chunk.

---

## Storage Architecture: Where Data Goes

To implement this, you need **two distinct storage systems** working together:

| Storage Type | What It Stores | Why It’s Used | Common Tools |
| --- | --- | --- | --- |
| **Vector Store** | Child Chunks + Embeddings + `parent_id` (in metadata) | For fast, semantic similarity searches. | Chroma, Pinecone, FAISS, Weaviate |
| **Docstore (Key-Value)** | Parent Chunks indexed by a unique `parent_id` | For rapid, exact-match retrieval of the full text once a child is found. | InMemoryStore, Redis, MongoDB, Local file system |

---

## Implementation Guide using LangChain

Here is how to set up and run a Parent Document Retriever in Python using LangChain.

### 1. Prerequisite Imports

```python
from langchain.retrievers import ParentDocumentRetriever
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import Chroma
from langchain_core.stores import InMemoryStore
from langchain_openai import OpenAIEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter

```

### 2. Define Your Text Splitters

You need two splitters: one to define your large parent contexts, and one to break those parents down into bite-sized children.

```python
# This text splitter is used to create the parent documents
parent_splitter = RecursiveCharacterTextSplitter(chunk_size=2000, chunk_overlap=200)

# This text splitter is used to create the child documents
child_splitter = RecursiveCharacterTextSplitter(chunk_size=400, chunk_overlap=50)

```

### 3. Initialize Your Storage Layers

Set up your vector database (for children) and your key-value document store (for parents).

```python
# The vectorstore to use to index the child chunks
vectorstore = Chroma(
    collection_name="split_parents", 
    embedding_function=OpenAIEmbeddings()
)

# The storage layer for the parent documents
docstore = InMemoryStore()

```

### 4. Create and Train the Retriever

LangChain provides a unified `ParentDocumentRetriever` class that manages the mapping between the two stores automatically.

```python
retriever = ParentDocumentRetriever(
    vectorstore=vectorstore,
    docstore=docstore,
    child_splitter=child_splitter,
    parent_splitter=parent_splitter,
)

```

### 5. Load Documents and Add to Retriever

When you call `add_documents`, LangChain automatically handles the heavy lifting: it splits the data into parents, splits those into children, links them via IDs, and stores them in their respective databases.

```python
# Load your data
loader = TextLoader("your_large_document.txt")
docs = loader.load()

# This handles splitting and storing in both vectorstore and docstore
retriever.add_documents(docs, ids=None)

```

### 6. Retrieve Full Context From Small Chunks

Now, when you invoke the retriever, it will search the vector store using the child chunks but return the parent chunks.

```python
# Search query
query = "What were the financial results for Q3?"

# This will return the large parent chunks associated with the matching child chunks
retrieved_docs = retriever.invoke(query)

print(f"Retrieved {len(retrieved_docs)} parent documents.")
print(retrieved_docs[0].page_content)

```

## Pro-Tip: "Full Document" Retrieval Strategy

If your original documents aren't monstrously huge and you want the LLM to see the *entire original file* whenever a small chunk matches, you can omit the `parent_splitter` entirely:

```python
# By leaving parent_splitter out, the raw, unsplit documents become the parents
retriever = ParentDocumentRetriever(
    vectorstore=vectorstore,
    docstore=docstore,
    child_splitter=child_splitter,
)

```

This ensures that if a single sentence inside a 5-page PDF matches the query, the LLM gets the entire 5-page context.
