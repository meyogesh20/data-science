# Project 2 : Test failure analysis using machine learning

Project 2 : Test failure analysis using machine learning
- Not all automation failures are real defects, but it could be due to several other reasons like environment stability issue or unavailability of any services or longer queue for batch jobs, setup issue or runtime issues, change in UI objects and their properties, change in functionality and new development.
- Teams tends to ignore results altogether as failure could have other issues.
- Engineers waste time in analyzing test failures that are not real defects.
- When the number of failures is high, its difficult to analyze the root cause of every test.

Problem complexity
- different teams use multiple framework and technical stack, depending upon framework the automation failure messages are different.
- also different programming languages and result logging formats are different. For example one testcase log format is .log which is text based, another one is xls format, one is XML format.
- Various test specification methodologies.
- Customized Error messages
- Failure could arise from multiple layers.
1. Test automation code
2. Libraries Used
3. Automation Framework used
4. Related to Test Run tools, e.g Selenium Web Driver, UFT Tool, Test complete
5. Programming Languages.

Solution Capabilities
- Our solution should able to
1. Automatically detect error pattern and classify them
2. Agnostic to different framework
3. Extensible for future framework
4. Ability to work with unstructured error message like stack trace.
5. Adaptable to new failures.
6. High Reliability -> Because this will determine the confidence with which the team would use the output of the solution.

Solution Outline
- We need machine learning techniques like NLP and various classification models to solve our problem statement.
- We need to classify automation failure error into the following categories.
1. Environment Issues - Test failures due to unstable environment (e.g a bad or slow service causing the application to be slow or behave erratically)
2. automation - Test failure due to automation only issues like clicking a non-existing element or not waiting for an element to exist/load properly.
3. Due to defect - Test failure due to real bug or defect in an application.

Plan -
1. Collect errors and classify them manually
2. Identify the correct algorithm
3. Proof of Concept (POC)
4. Create ML end-to-end flow
5. Automate everything.
6. Keep adding more pre-classified data and keep fine tuning.
7. consume the ML Model.
