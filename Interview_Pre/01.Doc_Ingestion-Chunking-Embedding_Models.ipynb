# **All major chunking strategies**

---

# 🔹 1. Fixed-size (naive) chunking

### ✅ What it is

* Split text into equal token/character sizes

### Example

* 500 tokens per chunk + 50 overlap

### 👍 Pros

* Simple, fast
* Works decently for uniform text

### 👎 Cons

* Breaks context mid-sentence
* Poor for structured docs

👉 Use when:

* Logs, raw text dumps

---

# 🔹 2. Recursive / hierarchical chunking

Used in frameworks like LangChain

### ✅ How it works

Split progressively:

```
Document → Sections → Paragraphs → Sentences
```

### 👍 Pros

* Preserves semantic boundaries
* Best general-purpose method

👉 Use when:

* PDFs, DOCX, articles

---

# 🔹 3. Sentence-based chunking

### ✅ What it is

* Group sentences until token limit reached

### Tools

* NLTK
* spaCy

### 👍 Pros

* Clean semantic chunks
* Good for QA tasks

👉 Use when:

* Natural language content

---

# 🔹 4. Semantic chunking (embedding-based)

### ✅ What it is

* Split where **meaning changes** using embeddings similarity

### How

* Compute embeddings for sentences
* Split when similarity drops

### 👍 Pros

* Most intelligent chunking
* Avoids topic mixing

### 👎 Cons

* Slower, more complex

👉 Use when:

* High-quality RAG (important systems)

---

# 🔹 5. Structure-aware chunking

### ✅ What it is

Chunk based on document structure:

* Headings
* Tables
* Sections

### 👍 Pros

* Preserves document logic
* Best for enterprise docs

👉 Use when:

* PDFs, DOCX, XML

---

# 🔹 6. Table-aware chunking

### ✅ What it is

* Treat tables separately

### Strategy

* Row-wise chunking
* Column-wise summarization

👉 Tools:

* pandas

👉 Use when:

* Excel, tables in PDFs

---

# 🔹 7. Sliding window (overlapping chunks)

### ✅ What it is

* Overlapping chunks to preserve context

Example:

```
Chunk 1: tokens 0–500  
Chunk 2: tokens 450–950
```

### 👍 Pros

* Avoids context loss

👉 Use ALWAYS with other strategies

---

# 🔹 8. Metadata-aware chunking

### ✅ What it is

Attach metadata:

```json
{
  "chunk": "...",
  "source": "test_case_1",
  "section": "expected_output"
}
```

### 👍 Pros

* Improves filtering + retrieval

---

# 🔹 9. Query-aware chunking (advanced)

### ✅ What it is

* Chunk based on expected queries
* Pre-group relevant info

👉 Use when:

* Domain-specific systems (like yours)

# **Clean Python implementations** for each major chunking strategy

---

# 🔹 1. Fixed-size chunking (with overlap)

```python
def fixed_size_chunk(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start = end - overlap  # overlap
    
    return chunks
```

---

# 🔹 2. Recursive / hierarchical chunking (like LangChain)

```python
def recursive_chunk(text, max_size=500):
    separators = ["\n\n", "\n", ".", " "]

    def split_text(text, sep_index=0):
        if len(text) <= max_size or sep_index >= len(separators):
            return [text]

        sep = separators[sep_index]
        parts = text.split(sep)

        chunks = []
        current = ""

        for part in parts:
            if len(current) + len(part) < max_size:
                current += part + sep
            else:
                chunks.extend(split_text(current.strip(), sep_index + 1))
                current = part + sep

        if current:
            chunks.extend(split_text(current.strip(), sep_index + 1))

        return chunks

    return split_text(text)
```

---

# 🔹 3. Sentence-based chunking

Using NLTK:

```python
import nltk
nltk.download('punkt')
from nltk.tokenize import sent_tokenize

def sentence_chunk(text, max_tokens=500):
    sentences = sent_tokenize(text)
    
    chunks = []
    current = ""
    
    for sent in sentences:
        if len(current) + len(sent) < max_tokens:
            current += " " + sent
        else:
            chunks.append(current.strip())
            current = sent
    
    if current:
        chunks.append(current.strip())
    
    return chunks
```

---

# 🔹 4. Semantic chunking (embedding-based)

Using cosine similarity:

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def semantic_chunk(sentences, embed_fn, threshold=0.75):
    embeddings = embed_fn(sentences)
    
    chunks = []
    current_chunk = [sentences[0]]
    
    for i in range(1, len(sentences)):
        sim = cosine_similarity(
            [embeddings[i-1]], [embeddings[i]]
        )[0][0]
        
        if sim < threshold:
            chunks.append(" ".join(current_chunk))
            current_chunk = [sentences[i]]
        else:
            current_chunk.append(sentences[i])
    
    if current_chunk:
        chunks.append(" ".join(current_chunk))
    
    return chunks
```

👉 `embed_fn` = your embedding model call

---

# 🔹 5. Structure-aware chunking (headings)

```python
import re

def structure_chunk(text):
    sections = re.split(r'\n(?=[A-Z][A-Z ]+\n)', text)
    return [sec.strip() for sec in sections if sec.strip()]
```

👉 Works well for:

```
INTRODUCTION
METHODS
RESULTS
```

---

# 🔹 6. Table-aware chunking (Excel/CSV)

Using pandas:

```python
import pandas as pd

def table_chunk(file_path):
    df = pd.read_excel(file_path)  # or read_csv
    
    chunks = []
    for _, row in df.iterrows():
        chunk = " | ".join([f"{col}: {row[col]}" for col in df.columns])
        chunks.append(chunk)
    
    return chunks
```

---

# 🔹 7. Sliding window chunking

```python
def sliding_window_chunk(text, window_size=500, stride=450):
    chunks = []
    
    for i in range(0, len(text), stride):
        chunk = text[i:i + window_size]
        chunks.append(chunk)
    
    return chunks
```

---

# 🔹 8. Metadata-aware chunking

```python
def metadata_chunk(chunks, source):
    return [
        {
            "content": chunk,
            "metadata": {
                "source": source
            }
        }
        for chunk in chunks
    ]
```

---

# 🔹 9. JSON/XML structure chunking

```python
import json

def json_chunk(json_data):
    chunks = []
    
    if isinstance(json_data, list):
        for item in json_data:
            chunks.append(json.dumps(item))
    elif isinstance(json_data, dict):
        for key, value in json_data.items():
            chunks.append(json.dumps({key: value}))
    
    return chunks
```

---

# 🔹 10. Log/event-based chunking

```python
import re

def log_chunk(log_text):
    events = re.split(r'\n(?=\d{4}-\d{2}-\d{2})', log_text)
    return [event.strip() for event in events if event.strip()]
```

👉 Assumes logs start with timestamps like:

```
2025-01-01 10:00:00 ERROR ...
```

---

# 🔹 Putting it all together (RAG pipeline)

```python
def process_document(text, doc_type):
    if doc_type == "pdf" or doc_type == "docx":
        chunks = recursive_chunk(text)
    elif doc_type == "log":
        chunks = log_chunk(text)
    elif doc_type == "json":
        chunks = json_chunk(text)
    else:
        chunks = fixed_size_chunk(text)
    
    return metadata_chunk(chunks, source=doc_type)
```
